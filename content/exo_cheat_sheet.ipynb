```python
# fft
import numpy as np
fs = 48000
x = [1, 2, 1, 2, 1, 2]
n = len(x)
X = np.fft.rfft(x, n)
f = np.fft.rfftfreq(n, 1/fs) # associated frequencies
print(X)
print(f)
```

```python
# ifft
import numpy as np
X = [0, 1, 5, 2]
n = len(X)
y = np.fft.irfft(X, n)
print(y)
```

```python
# create gate
import numpy as np
f = np.array([1, 2, 3, 4, 5, 6, 7, 8])
W = np.zeros(len(f))
W[f < 4] = 1
plt.plot(f, W);
```
