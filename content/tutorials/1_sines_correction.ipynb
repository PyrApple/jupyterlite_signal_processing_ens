# Correction

```python
import matplotlib.pyplot as plt
import numpy as np
import math
from IPython.display import Audio
```

## 1. Créer un vecteur temps

```python
# create time vector
duration_in_sec = 1
fs = 48000
n_sample = fs * duration_in_sec
t = np.arange(0, n_sample) / fs

# plot
plt.plot(t)
plt.xlabel('sample id'); plt.ylabel('time (sec)'); plt.title('time vector'); plt.show()

# plot zoom
selvect = np.arange(15)
plt.plot(t[selvect], '.')
plt.xlabel('sample id'); plt.ylabel('time (sec)'); plt.title('time vector (zoomed)'); plt.show()
```

## 2. Créer un sinus

```python
# create signal
fs = 48000
n_sample = duration_in_sec * fs
t = np.arange(0, n_sample) / fs
f1 = 100
x = np.sin(2*math.pi*f1*t)

# play
display(Audio(x, rate=fs))

# plot
plt.plot(t, x)
plt.xlabel('time (sec)'); plt.ylabel('amplitude'); 
plt.title(r'sin(2$\pi$ft), ' + f'f = {f1}Hz'); 
plt.show() 

# plot (zoom)
plt.plot(t, x)
# plt.stem(t, x, 'r')
plt.xlabel('time (sec)'); plt.ylabel('amplitude'); 
plt.title('zoom on a few periods'); 
plt.xlim([0, 3/f1])
plt.show()
```

## 3. Repliement spectral

```python
# create signal
f2 = 47000
# f2 = fs/2 + (fs/2 - f1)
x = np.sin(2*math.pi*f2*t)

# play
display(Audio(x, rate=fs))

# plot
plt.plot(t, x)
plt.xlabel('time (sec)'); plt.ylabel('amplitude'); 
plt.title(f'f2 = {f2}Hz'); 
plt.xlim([0, 3/f1])
plt.show() 
```
