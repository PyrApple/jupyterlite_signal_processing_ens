# Instructions

Connaître la réponse impulsionnelle d'un système permet de le caractériser complètement.  
Utiliser directement une impulsion est rarement envisageable (e.g. rupture membrane HP).  
On utilise souvent un sweep/chirp: un signal continu qui passe par toutes les fréquences. 

1. Créer un sweep 
    - à l'aide de la fonction `chirp` du module `scipy.signal`, créer un sweep logarithmique de $f_{start} = 100Hz$ à $f_{stop} = f_s/2$ avec $f_s = 48kHz$
    - le tracer, l'écouter
    - tracer son spectrogram à l'aide de la fonction `spectrogram` du module `scipy.signal`

Pour obtenir la réponse impulsionnelle d'une salle, on enregistre ce sweep dans une salle et on **déconvolue** cet enregistrement avec le sweep original.  
L'opération est généralement réalisée dans le domaine fréquentiel. 

2. Déconvolution
    - charger l'enregistrement "sweep_meas.wav" et le sweep original (anéchoïque) utilisé pour cet enregistrement "sweep_ref.wav", les écouter
    - filtrer le sweep mesuré avec un IIR passe-bande de type `butter` d'ordre 6 de fréquences $f_{low} = 50Hz$ et $f_{high} = 18kHz$
    - déconvolution: passer dans le domaine fréquentiel et diviser le sweep mesurée par celui de référence
    - création de la réponse impulsionnelle: repasser le résultat dans le domaine temporel, tracer, écouter

On utilise la même méthode pour égaliser la réponse en fréquence d'un HP ou d'un casque (la rendre "plate").  
Dans ce cas, l'inverse de la réponse en fréquence (mesurée avec le sweep) est utilisée pour pré-filtrer les sons joués sur ce HP / casque.
