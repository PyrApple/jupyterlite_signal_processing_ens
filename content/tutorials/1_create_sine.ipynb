# Objectif

1. Créer un vecteur temps $t = [0, 1/fs, 2/fs, ..., n\_samples/fs]$
2. Créer un sinus (vecteur) oscillant à $f_1 = 100Hz$, échantillonné à $fs = 48kHz$: le tracer et l'écouter
3. En exploitant le repliement spectral, faire osciller le sinus à $f_2 > fs$ de sorte que le pitch perçu soit le même que pour le sinus oscillant à $f_1$
