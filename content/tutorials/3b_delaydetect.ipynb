```python
# import 
import numpy as np

# local functions
def add_np_arrays(array1, array2):
    if( len(array2) > len(array1) ):
        array1 = np.pad(array1, [0, len(array2) - len(array1)])
    else:
        array2 = np.pad(array2, [0, len(array1) - len(array2)])
    return (array1 + array2) / 2
```
